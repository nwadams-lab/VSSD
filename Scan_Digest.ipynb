```python
from datetime import datetime


def calculate_security_score(results):
    udp_open = len(results["UDP_OPEN"])
    udp_of = len(results["UDP_OF"])
    tcp_open = len(results["TCP_OPEN"])

    score = 100
    score -= tcp_open * 10      
    score -= udp_open * 5      
    score -= udp_of * 2 

    score = max(0, min(100, score))
    return score


def write_security_report(results, score, timestamp, filename="security_report.txt"):
    with open(filename, "w") as f:
        f.write("=== VSSD Security Scan Report ===\n")
        f.write(f"Scan Time: {timestamp}\n\n")   # <-- timestamp added here
        f.write(f"Security Score: {score}/100\n\n")

        f.write("TCP Open Ports:\n")
        for port in results["TCP_OPEN"]:
            f.write(f"  - {port}\n")

        f.write("\nUDP Open Ports:\n")
        for port in results["UDP_OPEN"]:
            f.write(f"  - {port}\n")

        f.write("\nUDP Open|Filtered Ports:\n")
        for port in results["UDP_OF"]:
            f.write(f"  - {port}\n")

        f.write("\nReport generated by VSSD Scanner.\n")





from datetime import datetime

def run_full_security_scan():
    timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

    results = port_scan.full_scan()
    score = calculate_security_score(results)
    
    password = input("Enter password to check: ")
    password_results = password_checker.password_strength(password)
    password_score = password_results["score"]
    

    write_security_report(results, score, timestamp)

    print(f"Scan complete at {timestamp}. Security Score: {score}/100")

```
