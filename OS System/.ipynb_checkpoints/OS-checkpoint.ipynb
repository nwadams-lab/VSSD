```python
import platform
print(f'Platform: {platform.system()}')
plat = platform.system()
```

```python
#If user is using a Windows os, this gives them info on their system if it up to date or not
if plat == "Windows":
    current_ver = platform.version()
    latest_stab_ver = "10.0.26200"
    if current_ver < latest_stab_ver:
        print("System not up to date. Recommended to update your system.")
    elif current_ver == latest_stab_ver:
        print("System is up to date!")
    else:
        print("System exceeds common available version.")
#print(build)
```

```python
#If user is using a Linux os, this gives them info on their system if it up to date or not
if plat == "Linux":
    current_ver = platform.release()
    latest_stab_ver = "6.19.12"
    if current_ver < latest_stab_ver:
        print("System not up to date. Recommended to update your system.")
    elif current_ver == latest_stab_ver:
        print("System is up to date!")
    else:
        print("System exceeds common available version.") 
#Work in progress cannot test
```
